# **Step 2: Data Pre-processing**



## **2.1 Handling Missing Values**


```python
# Check for Missing Values in All Columns
print("Missing Values Summary:")
missing_values = df.isnull().sum()
print(missing_values)

print("="*70)

# Display Columns with Missing Values
if missing_values.sum() > 0:
    print("\nColumns with missing values:")
    print(missing_values[missing_values > 0])
else:
    print("\nNo missing values found in the dataset.")
```

```python
# Fill Missing Values in Categorical Columns
df['Item'] = df['Item'].fillna(
    df.groupby('Category')['Item'].transform(
        lambda x: x.mode()[0] if not x.mode().empty else 'Unknown'
    )
)
df['Discount Applied'] = df['Discount Applied'].fillna('Unknown')

print("Missing values after categorical cleaning:")
print(df[['Item', 'Discount Applied']].isnull().sum())

print("\nDiscount Applied value counts:")
print(df['Discount Applied'].value_counts())
```

```python
# Fill Missing Values in Numerical Columns
df['Price Per Unit'] = df['Price Per Unit'].fillna(df['Price Per Unit'].median())
df['Quantity'] = df['Quantity'].fillna(df['Quantity'].median())

# Recalculate Total Spent to keep it consistent
df['Total Spent'] = df['Price Per Unit'] * df['Quantity']
```

```python
print("Missing values after cleaning:")
print(df.isnull().sum())
```

```python
print("Rows where Total Spent is not equal to Price Per Unit × Quantity:")
print(((df['Price Per Unit'] * df['Quantity']) != df['Total Spent']).sum())
```

A consistency check was performed to verify whether Total Amount equals Price Per Unit multiplied by Quantity. This ensures that transaction totals are accurate after preprocessing.


## **2.2 Removing Duplicate Rows**



```python
# Check Duplicate Rows
df.duplicated().sum()
```

```python
# Remove Duplicate Rows
df.drop_duplicates(inplace=True)
```

## **2.3 Correcting Data Types**

```python
# Check Data Types
df.dtypes
```

```python
# Convert Transaction Date to Datetime
df['Transaction Date'] = pd.to_datetime(df['Transaction Date'], errors='coerce')

print("Invalid dates:", df['Transaction Date'].isnull().sum())
```

## **2.4 Column Renaming**

```python
# Rename Total Spent to Total Amount
df.rename(columns={'Total Spent': 'Total Amount'}, inplace=True)

print("Column renamed successfully!")
```

### **Visualization 12: Revenue Trend Over Time (Line Chart)**

**Objective**

Analyze changes in total revenue over time and identify sales trends, peaks, and fluctuations.

```python
# Convert Transaction Date to datetime
df['Transaction Date'] = pd.to_datetime(df['Transaction Date'])

# Create Year-Month column
df['Month'] = df['Transaction Date'].dt.to_period('M')

# Monthly Revenue
monthly_sales = df.groupby('Month')['Total Amount'].sum()

plt.figure(figsize=(12,5))

plt.plot(
    monthly_sales.index.astype(str),
    monthly_sales.values,
    marker='o'
)

plt.title('Monthly Revenue Trend')
plt.xlabel('Month')
plt.ylabel('Total Revenue')

plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

**Finding**

Monthly revenue shows noticeable fluctuations across the analysis period, indicating that sales performance varies from month to month. Certain months recorded significantly higher revenue, representing peak sales periods, while others experienced lower revenue levels, indicating slower business activity.

**Business Insight**

Identifying high-performing months helps the business plan inventory, staffing, and promotional campaigns more effectively. Low-performing months may require targeted marketing efforts to improve sales performance.
